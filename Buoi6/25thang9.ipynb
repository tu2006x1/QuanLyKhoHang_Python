```python
def chia(a, b):
    thuong = a // b
    du = a % b
    return thuong, du
```

```python
def tong(*args):
    s = 0
    for x in args:
        s += x
    return s

print(tong(10, 20))
print(tong(10, 20, 30, 40))
```

```python
def tinh_luong(gio, don_gia=50000):
    return gio * don_gia

tinh_luong(8)
```
