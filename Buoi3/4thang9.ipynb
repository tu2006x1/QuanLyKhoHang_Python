```python
#Bai 1.1
diem_so = [8.5, 7.0, 9.2, 6.5, 5.5]

print(diem_so[0])         # phan tu dau tien
print(diem_so[-1])        # phan tu cuoi cung
print(diem_so[1:4])       # cat tu vi tri 1 den truoc 4
print(diem_so[::2])       # lay cach 1 phan tu (step = 2)
print(diem_so[::-1])      # dao nguoc danh sach
```

```python
#Bai 1.2
ten_sv = ["An", "Binh", "Chi"]

ten_sv.append("Dung")       # them vao cuoi
print(ten_sv)
ten_sv.insert(1, "Em")      # chen vao vi tri 1
print(ten_sv)

ten_sv.remove("Chi")        # xoa theo gia tri
print(ten_sv)

pop_ra = ten_sv.pop()       # xoa va lay ra phan tu cuoi
print(ten_sv)

print(ten_sv, "- da xoa:", pop_ra)

ten_sv.sort()                    # sap xep tang dan (theo bang chu cai)
print(ten_sv)
ten_sv.reverse()                 # dao nguoc thu tu hien tai
print(ten_sv)

ten_sv.extend(["Giang", "Hoa"])  # noi them mot list khac vao
print(ten_sv)

```

```python
#Bai 2.1
diem_so = [8.5, 7.0, 9.2, 6.5, 5.5]
tong = 0

for diem in diem_so:
    print(diem)
    tong = tong + diem

print("Tong diem:", tong)
print("Diem trung binh:", round(tong / len(diem_so), 2))
```

```python
#Bai 2.2
ma_tran = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# In ra theo tung hang
for hang in ma_tran:
    print(hang)

# In ra tung phan tu, duyet theo hang roi theo cot
for hang in ma_tran:
    for phan_tu in hang:
        print(phan_tu, end=" ")
    print()

#viết thêm đoạn code tính tổng tất cả phần tử trong ma_tran bằng cách dùng biến tong cộng dồn qua 2 vòng for lồng nhau.
tong = 0
for hang in ma_tran:
    for phan_tu in hang:
        tong += phan_tu
print("Tong tat ca phan tu trong ma tran:", tong)
```

```python
#Bai 5
import math
diem_a = (2, 3)
diem_b = (7, 8)

xa, ya = diem_a
xb, yb = diem_b

khoang_cach = math.sqrt((xb - xa) ** 2 + (yb - ya) ** 2)
print(f"Khoang cach giua {diem_a} va {diem_b} la: {round(khoang_cach, 2)}")
cac_diem = [(0, 0), (3, 4), (6, 8)] # Tạo danh sách các tuple tọa độ

print("\nKhoang cach cua tung diem so voi goc toa do (0, 0):")
for diem in cac_diem:
    x, y = diem 
    kc_goc = math.sqrt(x**2 + y**2) #tính khoảng cách đến gốc
    print(f"- Diem {diem} cach goc toa do la: {round(kc_goc, 2)}")
```

```python
#Bai 6
danh_sach_sv = [(8.5, "An"), (7.0, "Binh"), (9.2, "Chi"), (6.5, "Dung")]


danh_sach_sv.append((8.0, "Em")) #thêm sv
danh_sach_sv.remove((7.0, "Binh")) #xóa sv

danh_sach_sv[0] = (9.0, danh_sach_sv[0][1]) #Sửa điểm sv
print("Chi co trong danh sach khong?", (9.2, "Chi") in danh_sach_sv) #kiểm tra sv có trong danh sách hay không

danh_sach_sv.sort() #tăng dần
print("Danh sach sau khi sap xep theo diem tang dan:")
for diem, ten in danh_sach_sv:
    print(f"{ten} - {diem}")


danh_sach_sv.sort(reverse=True) #giảm dần
print("Danh sach sau khi sap xep theo diem giam dan:")
for diem, ten in danh_sach_sv:
    print(f"{ten} - {diem}")
```

```python
#Bai 7
kho_hang = [
    ("Ban phim", 250000, 10),
    ("Chuot", 150000, 20),
    ("Man hinh", 2500000, 5),
]

kho_hang.append(("Tai nghe", 300000, 15)) #thêm sp
kho_hang.remove(("Chuot", 150000, 20)) #xóa sp
print("DANH SACH KHO HANG:")
for ten, gia, so_luong in kho_hang:
    print(f"{ten:<12} - Gia: {gia:>10,} - SL: {so_luong}")

tong_gia_tri = 0
for ten, gia, so_luong in kho_hang: #vòng lặp cộng dồn
    tong_gia_tri = tong_gia_tri + gia * so_luong

print(f"Tong gia tri kho hang: {tong_gia_tri:,} VND")
```
