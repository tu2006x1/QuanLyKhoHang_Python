```python
#Bài th2.2
# Nhập số nguyên dương n từ bàn phím
n = int(input("Nhập số nguyên dương n: "))

#Tính giai thừa của n (n!)
giai_thua = 1
for i in range(1, n + 1):
    giai_thua *= i
print(f"Giai thừa của {n} ({n}!) = {giai_thua}")

#Kiểm tra số nguyên tố
is_prime = True

if n < 2:
    is_prime = False
else:
    #Kiểm tra xem n có chia hết cho số nào từ 2 đến căn bậc hai của n không
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            is_prime = False
            break  # Thoát vòng lặp ngay nếu tìm thấy ước số

if is_prime:
    print(f"{n} là số nguyên tố.")
else:
    print(f"{n} không phải là số nguyên tố.")
```

```python
#Bài th2.3
num = int(input("Nhập một số nguyên dương: "))
tong = 0
temp = num  # Biến tạm để giữ giá trị gốc khi tính toán

while temp > 0:
    chu_so = temp % 10  # Lấy chữ số cuối cùng
    tong += chu_so  # Cộng vào tổng
    temp = temp // 10  # Loại bỏ chữ số cuối cùng (chia lấy nguyên)

print(f"Tổng các chữ số của {num} là: {tong}")
```

```python
#bài th3
n = int(input("Nhập số nguyên dương n: "))

# Bắt đầu kiểm tra từ số ngay sau n
so_tiep_theo = n + 1

while True:
    is_prime = True
    if so_tiep_theo < 2:
        is_prime = False
    else:
        for i in range(2, int(so_tiep_theo**0.5) + 1):
            if so_tiep_theo % i == 0:
                is_prime = False
                break  # Dừng kiểm tra số hiện tại nếu tìm thấy ước số
    
    if is_prime:
        print(f"Số nguyên tố đầu tiên lớn hơn {n} là: {so_tiep_theo}")
        break  # Thoát hẳn vòng lặp while khi đã tìm được số nguyên tố đầu tiên
        
    so_tiep_theo += 1
```

```python
#bài th3.1
danh_sach = [12, -5, 7, 0, 18, -3, 24, 9, 10]
print("Các phần tử hợp lệ (số dương chẵn) trong danh sách là:")

for num in danh_sach:
    # Nếu số nhỏ hơn hoặc bằng 0, hoặc là số lẻ -> Không hợp lệ
    if num <= 0 or num % 2 != 0:
        continue  # Bỏ qua các câu lệnh phía dưới, chuyển ngay sang phần tử tiếp theo
        
    # Chỉ những phần tử hợp lệ mới chạy đến dòng này
    print(num, end=" ")
print()
```

```python
#Bài th3.2
h = 5  # Chiều cao tam giác
for i in range(1, h + 1):
    for j in range(i):
        print("*", end="")
    print()

n = 5  # Số dòng của nửa hình thoi phía trên (phải là số nguyên dương)

# Nửa trên của hình thoi (gồm cả đường chéo giữa)
for i in range(1, n + 1):
    # In khoảng trắng phía trước
    for j in range(n - i):
        print(" ", end="")
    # In dấu sao
    for k in range(2 * i - 1):
        print("*", end="")
    print()

# Nửa dưới của hình thoi
for i in range(n - 1, 0, -1):
    # In khoảng trắng phía trước
    for j in range(n - i):
        print(" ", end="")
    # In dấu sao
    for k in range(2 * i - 1):
        print("*", end="")
    print()
```

```python
#Bài th4
import random

# 1. Máy chọn ngẫu nhiên một số từ 1 đến 100
so_bi_mat = random.randint(1, 100)
luot = 0
gioi_han_luot = 10

print("=== TRÒ CHƠI ĐOÁN SỐ ===")
print(f"Máy đã chọn một số từ 1 đến 100. Bạn có tối đa {gioi_han_luot} lượt đoán!")

# 2. Dùng while True để tạo vòng lặp vô hạn
while True:
    # Kiểm tra xem người chơi đã hết lượt chưa
    if luot >= gioi_han_luot:
        print(f"Rất tiếc, bạn đã hết lượt! Số bí mật là: {so_bi_mat}")
        break  # Dừng game khi hết lượt

    # Người chơi nhập số đoán
    doan = int(input(f"\nLượt {luot + 1}/{gioi_han_luot} - Nhập số đoán: "))
    luot += 1

    # 3. Dùng if - elif - else để kiểm tra và đưa ra gợi ý
    if doan == so_bi_mat:
        print(f"Chính xác! Bạn đã chiến thắng sau {luot} lượt.")
        break  # Dừng game ngay khi đoán đúng
    elif doan < so_bi_mat:
        print("Gợi ý: Lớn hơn")  # Số bí mật lớn hơn số vừa đoán
    else:
        print("Gợi ý: Nhỏ hơn")  # Số bí mật nhỏ hơn số vừa đoán
```
