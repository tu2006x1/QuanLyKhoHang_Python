```python
#Bai 2
diem_mon_hoc = {"Toan": 8.0, "Ly": 7.5, "Hoa": 9.0, "Van": 6.5}

for mon in diem_mon_hoc.keys():
    print(mon)

for diem in diem_mon_hoc.values():
    print(diem)

for mon, diem in diem_mon_hoc.items():
    print(f"{mon}: {diem}")

tong_diem = 0
for diem in diem_mon_hoc.values():
    tong_diem = tong_diem + diem

print("Diem trung binh:", round(tong_diem / len(diem_mon_hoc), 2))
```

```python
#Bai 3.1
diem_mon_hoc = {"Toan": 8.0, "Ly": 7.5, "Hoa": 9.0, "Van": 6.5}

diem_cong_diem = {mon: round(diem + 0.5, 2) for mon, diem in diem_mon_hoc.items()}
print(diem_cong_diem)

ten_mon_viet_hoa = {mon.upper(): diem for mon, diem in diem_mon_hoc.items()}
print(ten_mon_viet_hoa)

#Bài tập 3.2 – So sánh nhanh với Set:
mon_hoc_ky1 = {"Toan", "Ly", "Hoa", "Van"}
mon_hoc_ky2 = {"Toan", "Anh", "Tin", "Van"}

print(mon_hoc_ky1 & mon_hoc_ky2)   # giao: mon hoc chung 2 hoc ky
print(mon_hoc_ky1 | mon_hoc_ky2)   # hop: tat ca mon hoc ca 2 hoc ky
print(mon_hoc_ky1 - mon_hoc_ky2)   # mon chi co o hoc ky 1
```

```python
#Bài 5
dictionary = {
    "hello": "xin chào",
    "book": "quyển sách",
    "computer": "máy tính",
    "student": "sinh viên",
    "teacher": "giáo viên"
}

# Tra cứu từ
word = input("Nhập từ tiếng Anh cần tra: ")

if word in dictionary:
    print("Nghĩa tiếng Việt:", dictionary[word])
else:
    print("Từ này chưa có trong từ điển!")

# Thêm từ mới
word = input("\nNhập từ tiếng Anh mới: ")
meaning = input("Nhập nghĩa tiếng Việt: ")

dictionary[word] = meaning

print("\nTừ điển sau khi thêm:")
print(dictionary)

# Xóa từ
word = input("\nNhập từ cần xóa: ")

if word in dictionary:
    del dictionary[word]
    print("Đã xóa từ:", word)
else:
    print("Từ này không tồn tại!")

print("\nTừ điển cuối cùng:")
print(dictionary)
```

```python
#Bai 6. Vận dụng - Đếm tần suất từ trong văn bản
doan_van = "python la ngon ngu lap trinh python de hoc python de dung"

danh_sach_tu = doan_van.split()
tan_suat = {}

for tu in danh_sach_tu:
    tan_suat[tu] = tan_suat.get(tu, 0) + 1

print("Tan suat xuat hien cac tu:")
for tu, so_lan in tan_suat.items():
    print(f"{tu}: {so_lan}")
```

```python
#Hoạt động 7: Mini project – Quản lý điểm sinh viên bằng Dictionary 

quan_ly_diem = {
    "Nguyen Van A": [8.0, 7.5, 9.0],
    "Tran Thi B": [6.0, 6.5, 5.5],
    "Le Van C": [9.0, 9.5, 8.5],
}

# Them sinh vien moi
quan_ly_diem["Pham Thi D"] = [7.0, 8.0, 7.5]

# Sua diem mon dau tien cua mot sinh vien
quan_ly_diem["Tran Thi B"][0] = 7.0

diem_trung_binh = {}
for ho_ten, danh_sach_diem in quan_ly_diem.items():
    diem_trung_binh[ho_ten] = round(sum(danh_sach_diem) / len(danh_sach_diem), 2)

print("BANG DIEM TRUNG BINH:")
for ho_ten, dtb in diem_trung_binh.items():
    dat_loai_gioi = dtb >= 8.0
    print(f"{ho_ten:<15} - DTB: {dtb:<5} - Dat loai Gioi? {dat_loai_gioi}")
```
