```python
# Hoạt động 5.1

a = 17
b = 5

# Tính và in ra kết quả các toán tử số học
print("a + b =", a + b)    # Cộng
print("a - b =", a - b)    # Trừ
print("a * b =", a * b)    # Nhân
print("a / b =", a / b)    # Chia lấy số thập phân
print("a // b =", a // b)  # Chia lấy phần nguyên
print("a % b =", a % b)    # Chia lấy phần dư
print("a ** b =", a ** b)  # Lũy thừa (a mũ b)
```

```python
# Hoạt động 5.2
diem = 6.5
tuoi = 20

# 1. Kiểm tra điểm đạt loại Khá (từ 6.5 đến dưới 8.0)
loai_kha = (diem >= 6.5) and (diem < 8.0)
print("Điểm đạt loại Khá:", loai_kha)

# 2. Kiểm tra tuổi chưa đủ 18 hoặc trên 60
do_tuoi_dac_biet = (tuoi < 18) or (tuoi > 60)
print("Chưa đủ 18 hoặc trên 60 tuổi:", do_tuoi_dac_biet)

# 3. Phủ định lại điều kiện trên bằng toán tử not
phu_dinh_tuoi = not do_tuoi_dac_biet
print("Phủ định điều kiện tuổi (nằm trong khoảng 18 - 60):", phu_dinh_tuoi)
```

```python
# Hoạt động 5.3
# 1. Thực hành toán tử gán mở rộng (Cập nhật giá trị x liên tục)
x = 10
print("Giá trị ban đầu x =", x)

x += 5  # tương đương x = x + 5 (10 + 5)
print("Sau khi x += 5, x =", x)

x -= 3  # tương đương x = x - 3 (15 - 3)
print("Sau khi x -= 3, x =", x)

x *= 2  # tương đương x = x * 2 (12 * 2)
print("Sau khi x *= 2, x =", x)

x /= 4  # tương đương x = x / 4 (24 / 4)
print("Sau khi x /= 4, x =", x)

x //= 2  # tương đương x = x // 2 (6.0 // 2)
print("Sau khi x //= 2, x =", x)

x **= 3  # tương đương x = x ** 3 (3.0 mũ 3)
print("Sau khi x **= 3, x =", x)

print("-" * 30)

# 2. Thực hành toán tử đặc biệt (in, is)
danh_sach = [1, 2, 3, "python"]

# Dùng toán tử "in" kiểm tra số 3
co_trong_danh_sach = 3 in danh_sach
print("Số 3 có trong danh sách không?:", co_trong_danh_sach)

# Dùng toán tử "is" để so sánh 2 biến cùng tham chiếu
danh_sach_2 = danh_sach  # danh_sach_2 trỏ chung vào cùng 1 vùng nhớ với danh_sach
cung_tham_chieu = danh_sach_2 is danh_sach
print("Hai biến có cùng tham chiếu tới 1 list không?:", cung_tham_chieu)
```

```python
#Hoạt động 6.2
# 1. Gán cứng thông tin họ tên và điểm 3 môn học
ho_ten = "Nguyen Van A"
diem_toan = 8.0
diem_ly = 7.5
diem_hoa = 9.0

# 2. Tính điểm trung bình (Dùng toán tử số học)
dtb = (diem_toan + diem_ly + diem_hoa) / 3

# 3. Kiểm tra điều kiện xếp loại bằng biểu thức so sánh/logic
la_gioi = dtb >= 8.0
la_kha = dtb >= 6.5 and dtb < 8.0
la_trung_binh = dtb >= 5.0 and dtb < 6.5
la_yeu = dtb < 5.0

# 4. Xuất toàn bộ thông tin ra màn hình (Hàm round(dtb, 2) giúp làm tròn 2 chữ số thập phân)
print(ho_ten, "- DTB:", round(dtb, 2))
print("Dat loai Gioi?:", la_gioi)
print("Dat loai Kha?:", la_kha)
print("Dat loai Trung binh?:", la_trung_binh)
print("Dat loai Yeu?:", la_yeu)
print("Kieu du lieu cua la_gioi:", type(la_gioi))
```
